## 0. 先认识终端这个「黑窗口」

### 0.1 为什么躲不开它

图形界面（点鼠标）适合探索，但工程师的日常操作大多发生在终端里，因为：

- **可复现**：一条命令就是一步操作，写进文档别人照敲就行，不会出现「点左边第三个按钮」这种描述；
- **可自动化**：命令可以写进脚本批量执行，鼠标点击不能；
- **远程服务器只有终端**：训练 RL 策略常常要连实验室服务器 / 云主机，那里**没有图形界面**。

本课程里你会反复用到这些终端操作：

| 课程环节 | 要用到的命令 |
|----------|-------------|
| 安装/同步依赖 | `uv sync` |
| 启动 Jupyter | `uv run jupyter lab` |
| 跑单元测试 | `uv run pytest` |
| 查看训练产物 | `ls runs/`、`cat runs/.../train.log` |
| 训练卡死时救场 | `Ctrl+C`、`ps` + `kill` |

### 0.2 打开终端

- **Windows**：按 `Win + R` 输入 `cmd` 回车（CMD）；或开始菜单搜「PowerShell」；Win11 推荐「终端 Terminal」App。
- **Linux**：按 `Ctrl + Alt + T`，或在应用列表里找 Terminal。
- **macOS**：`Cmd + 空格` 搜「终端 / Terminal」。

### 0.3 读懂提示符

打开后那一行字叫**提示符（prompt）**，它在告诉你「我在哪个目录、等你输入」：

```
# Windows CMD：            当前路径 + 大于号
C:\Users\wangf>

# Linux / macOS：          用户名@主机名:当前目录 + $（# 表示 root 超级用户，要格外小心）
wangf@ubuntu:~/robot_rl_learn$
```

你敲的命令跟在提示符后面，回车执行。**本文所有命令示例都不包含提示符本身**，照敲即可。

## 场景 3 · 走来走去 —— `cd`

`cd`（change directory）是在目录之间移动的唯一方式。掌握四个「坐标」就够用了：

| 写法 | 含义 | Windows 示例 | Linux 示例 |
|------|------|-------------|-----------|
| `cd 名字` | 进入当前目录下的子目录 | `cd runs` | `cd runs` |
| `cd ..` | 返回上一级目录 | `cd ..` | `cd ..` |
| `cd 绝对路径` | 一步到位去哪都行 | `cd C:\Users\wangf\robot_rl_learn` | `cd /home/wangf/robot_rl_learn` |
| `cd ~` / `cd` | 回家目录 | `cd %USERPROFILE%` | `cd ~` 或直接 `cd` |

**三个保命技巧：**

1. **Tab 补全**：敲目录名的前几个字母按 `Tab`，终端自动补全，既快又防拼错（CMD/PowerShell/bash 都支持）；
2. **路径含空格要加引号**：`cd "C:\Program Files"` / `cd "/home/wangf/my folder"`；
3. **方向键 ↑**：调出上一条命令，不用重敲。

> 相对路径 vs 绝对路径：相对路径是「从当前位置出发怎么走」（`cd ../src`），
> 绝对路径是「不管在哪都通用的完整地址」（`cd /home/wangf/robot_rl_learn/src`）。
> 写脚本、写文档时优先绝对路径，不容易出错。

````python
# notebook 里用 %cd 魔法命令切换（终端里用 cd，效果相同）
import os
from pathlib import Path

NB_DIR = Path(os.getcwd())              # 本 notebook 所在目录 notebooks/00_basic
PROJECT_ROOT = NB_DIR.parent.parent     # 项目根目录
%cd {PROJECT_ROOT}
!pwd
!ls
````

## 场景 4 · 搭建与拆除 —— 新建/删除目录和文件

| 操作 | Windows CMD | Linux / macOS |
|------|-------------|---------------|
| 新建目录 | `mkdir my_lab` | `mkdir my_lab` |
| 新建多级目录 | `mkdir a\b\c` | `mkdir -p a/b/c` |
| 新建空文件 | `type nul > empty.txt` | `touch empty.txt` |
| 删除文件 | `del old.txt` | `rm old.txt` |
| 删除空目录 | `rmdir my_lab` | `rmdir my_lab` |
| 删除非空目录 | `rmdir /s my_lab` | `rm -r my_lab` |

> ⚠️ **删除命令没有回收站！** `del` / `rm` 删掉的文件基本找不回来。
> 尤其警惕 `rm -rf 目录`（强制递归删除，不询问）：多敲一个空格、变量为空，
> 都可能把工作成果甚至系统文件抹掉。执行删除前，先 `ls` / `dir` 确认目标，再敲删除命令。

## 场景 9 · 乾坤大挪移 —— 重定向与管道

命令的输出默认打在屏幕上，但你可以把它**导到别处**：

| 操作 | 写法（Windows 与 Linux 通用！） | 含义 |
|------|------|------|
| 输出覆盖写入文件 | `命令 > out.txt` | 屏幕输出存进文件（原内容被清空） |
| 输出追加到文件 | `命令 >> out.txt` | 接在文件末尾（不破坏原内容） |
| 管道串联 | `命令A \| 命令B` | A 的输出作为 B 的输入 |

```bash
# Linux 示例（Windows CMD 同样支持 > 和 |）
python train.py > runs/train_output.txt        # 训练输出存文件，后台慢慢看
grep INFO train.log | wc -l                    # 数一下日志里有多少条 INFO
```

> ⚠️ `>` 会**清空**目标文件再写入。想把新日志追加到旧日志后面，一定用 `>>`。
> 课程实战：`uv run pytest > test_result.txt 2>&1` 把测试结果（含报错）完整存档，方便发给助教提问。
> 其中 `2>&1` 表示「错误输出也一起进文件」。

````python
!grep "INFO" logs/train.log > info_only.txt
!cat info_only.txt
print("-" * 40)
!cat logs/train.log | grep INFO | wc -l
````

## 场景 10 · 隐形配置 —— 环境变量

环境变量是「终端全局可见的键值对配置」。最重要的一个叫 `PATH`：
它是一份**目录清单**，你敲 `python` 时，终端按顺序去这些目录里找名叫 `python` 的可执行文件。

| 操作 | Windows CMD | Linux / macOS |
|------|-------------|---------------|
| 查看某个变量 | `echo %PATH%` | `echo $PATH` |
| 临时设置（仅当前窗口） | `set DEMO=hello` | `export DEMO=hello` |
| 查看全部变量 | `set` | `env` |

> 常见坑：「明明装了 Python，终端却说不认识」——多半是安装目录没加进 `PATH`。
> 本课程用 `uv run` 规避了大部分这类问题。
> 注意：`set` / `export` 设置的变量只在**当前终端窗口**有效，关掉窗口就没了（这通常是好事）。

````python
import os

%env DEMO_VAR=hello_rl
!echo $DEMO_VAR
print("-" * 40)
# PATH 很长，只看前两个目录，感受一下它的结构（冒号分隔）
print("\n".join(os.environ["PATH"].split(":")[:2]))
````

## 场景 11 · 收拾残局 —— 查看与结束进程

训练卡死、Jupyter 关不掉、程序占着 GPU 不放——都需要进程管理。

| 操作 | Windows CMD | Linux / macOS |
|------|-------------|---------------|
| 列出进程 | `tasklist` | `ps aux` |
| 按名字过滤 | `tasklist \| findstr python` | `ps aux \| grep python` |
| 结束进程（按 PID） | `taskkill /PID 12345 /F` | `kill 12345` |
| 强制结束 | 同上（`/F` 即强制） | `kill -9 12345`（最后手段） |

每个进程有一个编号 **PID**（Process ID）。结束进程的标准动作：先查到 PID，再 `kill` 它。

> 温柔 vs 强制：Linux 的 `kill PID` 是「请你退出」（进程有机会保存现场）；
> `kill -9 PID` 是「直接枪毙」，可能导致文件写一半损坏。**先礼后兵**。
> 在终端里跑的前台程序，优先用 `Ctrl+C` 中断，轮不到 kill 出场。

````python
import subprocess

# 启动一个「睡 60 秒」的进程充当训练卡死的演员
p = subprocess.Popen(["sleep", "60"])
print("启动进程 PID =", p.pid)
!ps aux | grep "[s]leep 60" | head -2
````

````python
!kill {p.pid}          # 按 PID 结束它（Windows: taskkill /PID {p.pid} /F）
p.wait()
print("进程已结束，退出码:", p.returncode)
````

## 小结

- 终端三问：**我在哪（`pwd`）、有什么（`ls`）、怎么去（`cd`）**——先解决认路问题；
- 操作文件五件套：`mkdir / cp / mv / rm / cat`，其中 `rm` 和 `>` 是「不可撤销」操作，动手前先确认；
- 查日志靠 `tail` 和 `grep`，跑程序靠 `命令 + 参数`，救场靠 `Ctrl+C` 和 `kill`；
- Windows 与 Linux 命令名字不同，**概念完全相通**——学会一套，另一套对照速查表十分钟上手；
- 记不住参数不丢人，`命令 --help` / `man` / 问 AI，是工程师的日常。
